```python
# distance from origin  -> eucliden norm

import numpy as np 

A = np.array([1,2,3,4,5])

# calculate the eucliden destance from origin(L2)

distance = np.linalg.norm(A)

distance
```

```python

# Euclidean Distance in n-D space

A = np.array([1,2,3,4,5])
B = np.array([6,7,8,9,10])

# calculate the diffrence  vector
difference = A - B
difference
```

```python
# calculate the Euclidean distance between A and B (L2 norm of the distance)

distance = np.linalg.norm(difference)

print("Eucliden distance between A and B :", distance)
```

```python
A = np.array([1,2,3])
B = np.array([-4,-2,-5])
C = np.array([5,3,3])

cosine_similarity = np.dot(A,B) / (np.linalg.norm(A)* np.linalg.norm(B)) 

print("cosine similarity b/w points A and C :",cosine_similarity)

# calculate  same above cosine similarity between point to chek how much they similar to each other its used in like move recoomendarion system

cosine_similarity1 = np.dot(A,C) / (np.linalg.norm(A) * np.linalg.norm(C))

print("cosine similarity between B and C is :", cosine_similarity1)
```
